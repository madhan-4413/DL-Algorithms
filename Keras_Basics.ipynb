## 3. Data Preparation
For this demonstration, we will create a simple dummy dataset using NumPy. This dataset will consist of random numbers for both training and testing features (`X_train`, `X_test`) and their corresponding labels (`y_train`, `y_test`).

```python
# Dummy Dataset
import numpy as np

#Random Dataset
X_train=np.random.random((1000,20))
y_train=np.random.random((1000,1))


X_test=np.random.random((100,20))
y_test=np.random.random((100,1))
```

## 4. Model Definition
We will define a simple Sequential model, which is a linear stack of layers. Our model will consist of fully connected (Dense) layers.

# Creating a Model

## Layers


*   Dense Layer -> A layer of fully connected neurons
*   Dropout
*   Conv1D, conv2D,
*   LSTM/GRU



## Install required libraries

```python
# Imports
from keras.models import Sequential
from keras.layers import Dense

from keras import optimizers
from keras import metrics
```

### 4.1 Importing Necessary Libraries
First, we import `Sequential` for building the model and `Dense` for creating fully connected layers from Keras.

```python
# define the structure of the model
model=Sequential()

# Input Layer / Hidden Layer
model.add(Dense(10,activation='relu',input_dim=X_train.shape[1]))

# Second Layer
model.add(Dense(10,activation='relu'))

# Output Layer
model.add(Dense(1,activation='relu'))
model.summary()
```

The model summary above shows the layers, output shape, and the number of parameters for each layer. The `None` in the output shape indicates that the layer can handle any batch size.

```python
# Configure the model for learning/training
# Loss, metrics, optimizer

model.compile(loss='mse',
              optimizer='adamax',)
              # metrics=['accuracy'])
```

## 6. Model Training
Now, we will train our model using the `fit` method. We'll specify the training data (`X_train`, `y_train`), `batch_size`, and the number of `epochs`.

```python
# Training the model
model.fit(X_train,y_train,batch_size=500,epochs=30)
```

## 7. Model Prediction
After training, we can use the `predict` method to get predictions on new, unseen data (`X_test`).

```python
y_pred=model.predict(X_test)
```

### Prediction Output
The `y_pred` variable now holds the predictions made by the trained model on the `X_test` dataset.
