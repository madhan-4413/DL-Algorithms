1. ## Tensors of Different Ranks
2. **Purpose**: Demonstrates the creation of tensors with varying dimensions.

```python
import tensorflow as tf

# 0D Tensor (scalar)
t0 = tf.constant(7)

# 1D Tensor (vector)
t1 = tf.constant([1, 2, 3])

# 2D Tensor (matrix)
t2 = tf.constant([[1, 2],
                  [3, 4]])

# 3D Tensor (cube / stack of matrices)
t3 = tf.constant([[[1],[2]], [[3],[4]]])

print(t0.shape, t1.shape, t2.shape, t3.shape)
```



3. **Explanation**: This cell creates and showcases tensors of different ranks (dimensions):
    - **0D Tensor (Scalar)**: `t0` (single value).
    - **1D Tensor (Vector)**: `t1` (sequence).
    - **2D Tensor (Matrix)**: `t2` (grid).
    - **3D Tensor**: `t3` (stack of matrices).
   The `print(t0.shape, ...)` output clearly illustrates how the `.shape` attribute indicates the dimensionality of each tensor.

1. ## Initializing 2D Tensors
2. **Purpose**: Sets up two 2x2 matrices for tensor operations.

```python
x = tf.constant([[1.0, 2.0],
                 [3.0, 4.0]])

y = tf.constant([[10.0, 20.0],
                 [30.0, 40.0]])
```



3. **Explanation**: This cell defines two new 2-dimensional tensors, `x` and `y`, which are essentially 2x2 matrices containing floating-point numbers. These tensors serve as inputs for exploring various arithmetic and manipulation operations in the following cells.

1. ## Element-wise Tensor Operations
2. **Purpose**: Performs basic arithmetic on tensors, element by element.

```python


# Element-wise operations
print("Add:\n", x + y)
print("Multiply:\n", x * y)
print("Subtract:\n", x - y)
```


3. **Explanation**: This cell demonstrates element-wise addition, multiplication, and subtraction between tensors `x` and `y`. In these operations, corresponding elements from each tensor are combined individually, resulting in a new tensor of the same shape where each element is the result of the operation on its respective inputs.

1. ## Matrix Multiplication
2. **Purpose**: Executes the fundamental linear algebra operation of matrix multiplication.

```python

# Matrix multiplication
print("MatMul:\n", tf.matmul(x, y))
```


3. **Explanation**: Unlike element-wise operations, `tf.matmul(x, y)` performs true matrix multiplication. This operation is crucial in many machine learning algorithms. The result is a new matrix where each element is computed by taking the dot product of a row from the first matrix (`x`) and a column from the second matrix (`y`).

1. ## Tensor Reductions and Reshaping
2. **Purpose**: Illustrates how to aggregate tensor data and change its structural dimensions.

```python

# Reductions
print("Sum:", tf.reduce_sum(x))
print("Mean:", tf.reduce_mean(x))

# Reshape
print("Reshape:\n", tf.reshape(x, (4,)))
```


3. **Explanation**: This cell explores two key tensor manipulations:
    - **Reductions**: `tf.reduce_sum()` and `tf.reduce_mean()` compute aggregate statistics (sum, average) across all elements of a tensor, reducing its dimensionality to a scalar.
    - **Reshaping**: `tf.reshape(x, (4,))` modifies the tensor's shape from a 2x2 matrix to a 1D vector of 4 elements. This operation reinterprets the tensor's elements into a new arrangement without altering the underlying data.

1. ## Simple Linear Regression with Gradient Descent
2. **Purpose**: Implements a basic machine learning model to find a linear relationship in data.

```python
import tensorflow as tf

# Training data (X=inputs, y=targets)
X = tf.constant([[2.0], [4.0], [6.0], [8.0]])
y = tf.constant([[4.0], [8.0], [12.0], [16.0]])

# Trainable parameters (weights & bias)
w = tf.Variable(0.0)
b = tf.Variable(0.0)

lr = 0.01
epochs = 200

for epoch in range(epochs):
    with tf.GradientTape() as tape:
        y_pred = X * w + b
        loss = tf.reduce_mean((y - y_pred)**2)

    dw, db = tape.gradient(loss, [w, b])

    w.assign_sub(lr * dw)
    b.assign_sub(lr * db)

    if epoch % 40 == 0:
        print(f"Epoch {epoch}, Loss={loss.numpy():.4f}, w={w.numpy():.4f}, b={b.numpy():.4f}")

print("\nFinal parameters:")
print("w =", w.numpy(), " b =", b.numpy())
print("Prediction for X=10:", (10*w + b).numpy())
```


3. **Explanation**: This cell demonstrates a simple linear regression model using TensorFlow. It defines input `X` and target `y` data, then iteratively adjusts `w` (weight) and `b` (bias) using **gradient descent** to minimize the prediction error (loss). A `tf.GradientTape` automatically calculates the gradients needed to update `w` and `b`, guiding the model towards the best fit. Finally, it shows the optimized parameters and a prediction for a new input.
